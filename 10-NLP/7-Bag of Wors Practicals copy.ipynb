```python
import pandas as pd

data = {
    "label": [
        "ham","ham","spam","ham","spam","ham","spam","ham","spam","ham",
        "spam","ham","ham","spam","ham","spam","ham","spam","ham","spam",
        "ham","ham","spam","ham","spam","ham","spam","ham","spam","ham"
    ],
    "message": [
        "Hey, are we meeting at 6 PM today?",
        "Don't forget to bring your laptop tomorrow.",
        "Congratulations! You have won a FREE iPhone. Click here now.",
        "Can you send me the notes after class?",
        "Limited time offer! Claim your cash reward today.",
        "I'm on my way home.",
        "You have been selected for a vacation worth $5000. Call now!",
        "Happy Birthday! Have a wonderful day.",
        "URGENT! Your account has been suspended. Verify immediately.",
        "Let's play cricket this evening.",
        "Win exciting prizes by texting WIN to 80808.",
        "Please submit the assignment before Friday.",
        "I'll call you after lunch.",
        "Earn money from home without investment. Join today!",
        "Can you pick up some groceries?",
        "Exclusive deal! Buy one get one free. Shop now!",
        "The meeting has been postponed to tomorrow.",
        "Your mobile number has won a lucky draw. Claim now.",
        "Where are you? Everyone is waiting.",
        "Congratulations! You've won ₹50,000. Reply YES to claim.",
        "Let's go for coffee after work.",
        "Thanks for helping me yesterday.",
        "Lowest loan rates available. Apply instantly.",
        "Can you share today's timetable?",
        "FREE recharge worth ₹100. Click the link below.",
        "Dinner is ready. Come downstairs.",
        "You are pre-approved for a credit card. Apply today.",
        "Good luck for your exam tomorrow!",
        "Exclusive membership offer expires tonight!",
        "See you at the library."
    ]
}

messages = pd.DataFrame(data)

messages
```

```python
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
from nltk.stem import WordNetLemmatizer
import re
```

```python
sn=SnowballStemmer('english')
lm=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(len(messages)):
    review=re.sub('[^a-zA-Z]'," ",messages['message'][i])
    review=review.lower().split()
    review=[sn.stem(word) for word in review if word not in set(stopwords.words("english"))]
    review=" ".join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=100)
X = cv.fit_transform(corpus)
```

```python
X = cv.fit_transform(corpus).toarray()
X
```
