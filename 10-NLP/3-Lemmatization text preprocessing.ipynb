```python
import nltk

# Download the specific missing resource
nltk.download('wordnet')

# Optional: If you are using lemmatization, you often need the omw-1.4 dataset as well
nltk.download('omw-1.4') 
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
Lemma=WordNetLemmatizer()
```

```python
Lemma.lemmatize("Going",pos="v")
```

```python
for word in words:
    print(word+"---->"+Lemma.lemmatize(word,pos="v"))
```

```python
Lemma.lemmatize("fairly",pos="n"),Lemma.lemmatize("sportly",pos="r")
```
