```python
## Tokenization
### Sentence -->Paragraph
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
from nltk.tokenize import wordpunct_tokenize
from nltk.tokenize import TreebankWordTokenizer


```

```python
print(corpus)
```

```python
x=sent_tokenize(corpus)
print(x)
for i in x:
    print(i)

```

```python
## Tokenization
### Paragraph --> Words
```

```python
y=word_tokenize(corpus)

for i in y:
    print(i)
```

```python
x=sent_tokenize(corpus)
print(x)
for i in x:
    print(word_tokenize(i))
```

```python
document=wordpunct_tokenize(corpus)
```

```python
print(document)
```

```python
x=TreebankWordTokenizer()
x.tokenize(corpus)
```
