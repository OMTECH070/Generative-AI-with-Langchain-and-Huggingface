```python
words=["eaten","eat","eating","sleep","slept","sleeping","program","programming","programmer","stop","stoping", "walk","walking"]
```

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
Poter=PorterStemmer()
for word in words:
    print(word+"---->"+Poter.stem(word))
```

```python
from nltk.stem import RegexpStemmer
```

```python
Rege=RegexpStemmer('ing$|s$|es$|able$', min=4)

for word in words:
    print(word+"---->"+Rege.stem(word))
```

```python
from nltk.stem import SnowballStemmer
```

```python
Snow=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+Snow.stem(word))
```
