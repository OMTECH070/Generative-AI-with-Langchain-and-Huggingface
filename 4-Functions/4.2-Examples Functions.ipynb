## Example 1: Temperature Conversion

```python
def tempC(value):
    x=(9/5)*value+32
    return x

def tempF(value):
    x=(value-32)*5/9
    return x
```

```python
print(tempC(37))
print(tempF(98.8))
```

## Example 2: Password Strength Checker

```python
def passw(password):

    if len(password)<8:
        return False
    if not any (char.isdigit() for char in password):
        return False
    if not any (char.lower() for char in password):
        return False
    if not any (char.isupper() for char in password):
        return False
    if not any (char in "!@#$%^&()_+" for char in password):
        return False
    return True

print(passw("Wakvhdv"))
print(passw("weaksT5$"))


        
```

## Example 3: Calculate the Total cost of item in a shopping cart

```python
def calculate(carts):
    total=0
    for cart in carts:
        print(cart)
        total+=cart["Price"]*cart["Quantity"]
        
    return total




cart=[{"Name":"Apple","Price":200,"Quantity":20},
{"Name":"Orange","Price":20,"Quantity":200},
{"Name":"Kiwi","Price":300,"Quantity":20}]

calculate(cart)
```

## Example 4: Check if string is Palindrome

```python
def line(string):
    if string[::-1]==string:
        print("The string is Palindrome")
    else:
        print("The string is not Palindrome")
    


line("nitin")
```

## Example 5: Calculate the Factorial of a number using recursion

```python
def factorial(x):
    if x==0:
        return 1
    else:
        z = x*factorial(x-1)
        print(z)
        return z
factorial(5)
    
```

## ## Example 6: A Function To read a File and count the frequency of each word

```python
def count_word_frequency(file_path):
    word_count={}
    with open(file_path,"r") as file:
        for line in file:
            words=line.split()
            for word in words:
                word=word.lower().strip('!?:;",')
                word_count[word]=word_count.get(word,0)+1
    return word_count

file_path="sample.txt"
frequency=count_word_frequency(file_path)
print(frequency)
```
