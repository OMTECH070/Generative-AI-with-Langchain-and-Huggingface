## Map Function

```python
numbers=[1,2,3,4,5,6,7,8]

list(map(lambda x:x**2,numbers))
```

```python
##Multiple iterables

num1=[1,2,3]
num2=[4,5,6]

added_num=list(map(lambda x,y :x+y ,num1,num2))
print(added_num)
```

```python
## map() to convert a list of strings to integers

str_num=["1","2","3","4","5"]
int_num=list(map(int,str_num))

print(int_num)
print(type(int_num))
```

```python
words=["Apple","Banana","Cherry"]
lower=list(map(str.lower,words))
print(lower)
```

```python
def get_name(person):
    return person["Name"]

people=[
    {"Name":"Om","Age":19},
    {"Name":"Sai","Age":12}
]

list(map(get_name,people))
```
