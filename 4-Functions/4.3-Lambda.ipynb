## Lambda function

syntax

lambda argument:expression

```python
def add(a,b):
    return a+b
```

```python
add(2,6)
```

```python
add=lambda a,b:a+b
print(add(5,9))

```

```python
def even(num):
    if num%2==0:
        return True

even(24)
```

```python
even=lambda num: num%2==0 

print(even(29))
```

```python
##map 
number=[1,2,3,4,5,6]
def square(number):
    lst=[]
    for num in number:
        lst.append(num**2)
    return lst  

print(square(number))      
```

```python
list(map(lambda x:x**2,number))
```
