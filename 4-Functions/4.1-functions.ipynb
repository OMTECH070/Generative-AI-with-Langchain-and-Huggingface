```python
num=25

if num%2==0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
def num(x):
    """This function finds even or odd"""
    if x%2==0:
        return print(f"The number {x} is even")
    else:
        return print(f"The number {x} is odd")

num(56)
num(55)
num(32)

```

```python
##Default Parameters
def greet(name="Guest"):
    print(f"Hello {name}")

greet("Om")
```

```python
###Variable length arguments
##Positional arguments

def print_number(*arg):
    for number in arg:
        print(number)
```

```python
print_number(1,2,3,4,5,6,7,8,9,"Om")
```

```python
##Keyword arguments
def print_detail(**kargs):
    for key,values in kargs.items():
        print(f"{key}:{values}")
```

```python
print_detail(name="Om",age=18)
```

```python
##Keyword arguments and positional
def print_detail(*args,**kargs):
    for val in args:
        print(f"Positional argu:{val}")
    for key,values in kargs.items():
        print(f"{key}:{values}")
```

```python
print_detail(1,2,3,4,5,865,name="Om",age=465465)
```
