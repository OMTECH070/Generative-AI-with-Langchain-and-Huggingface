## File Operation- Read and Write files

```python
### Read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
## Writing a file(Overwriting)

with open('example.txt','w') as file:       
    file.write("Hello world\n")
    file.write("This is a new line\n")

```

```python
## Writing a file(Without Overwriting)

with open('example.txt','a') as file:
    file.write("\nHello world vuiuibvuer\n")
       
```

```python
## Writing a list of lines to a file
lines=["First line\n","Second line\n","Third line\n"]
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
## Read the content from source text file and write to a destination text file

with open("example.txt","r") as file:
    content=file.read()
    print(content)

with open("example1.txt","a") as file:
    file.write(content)
    
```

```python
##Counting lines,words and charaters in a text file

def counting(file_path):
    lst=[]
    count={}
    with open(file_path,"r") as file:
        lines=file.readlines()
        num_Lines=len(lines)
        num_words=sum(len(line.split()) for line in lines)
        num_char=sum(len(line) for line in lines)
        for line in lines:
            word=line.lower().strip('!?,.":;').split()
            for w in word:
                lst.append(w)
        for word in lst:
            if word in count:
                count[word]+=1
            else:
                count[word]=1



    return num_Lines,num_words,num_char,count

lines,words,char,count=counting("example.txt")
print(f"lines:{lines} , Words:{words} , Charaters:{char}")
print(count)
```

```python
## Writing in a file and reading the file

with open ("example1.txt","w+") as file:
    file.write("Hello world\n")
    file.write("This is the 2nd line\n")

    file.seek(0)

    content=file.read()
    print(content)
```
