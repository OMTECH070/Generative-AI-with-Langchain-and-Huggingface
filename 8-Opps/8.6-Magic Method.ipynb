## Magic Method

Magic method in python,also known as dunder methods(doubbble underscore methods),are special methods that start and end with double underscores.These enables you to define the behavior of the objects for built-in-operation,such as arithmetic operation,comparison and more

```python
"""
__init__ : Initializes a new instance of a class.
__str__ : Returns a string representation of an object.
__repr__ : Returns an official string representation of an object.
__len__ : Returns the length of an object.
__getitem__ : Gets an item from a container.
__setitem__ : Sets an item in a container.
"""
```

```python
class Person:
    pass
person=Person()
dir(person)
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

person=Person("Om",19)
print(person)
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def __str__(self):
        return f"{self.name},{self.age} years old"

person=Person("Om",19)
print(person)
```
