## Encapsulation

The concept of wrapping data and methods together as a single umit.It restricts direct access to some of the object's components,which is a means of preventing accidental interference and misuse of the data

```python
## Encapsulation with getter and setter method

### Public,protected,private variables or access modifiers

class Person:
    def __init__(self,name,age):
        self.name=name ## public variable 
        self.age=age   ## public variable 

def get_name(person):
    return person.name

person=Person("Om",19)
print(person.name)
get_name(person)
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.__name=name            ## private variable       -This variable cannot be accessed outside the class nor by the derived class
        self._age=age               ## proctected variable    -This variable cannot be accessed outside the class but can be accessed by derived class
        self.gender=gender          ## punliv variable        -This variable can be accessed by anyone

class Employee(Person):
    def __init__ (self,name,age,gender):
        super().__init__(name,age,gender)



```

```python
person=Employee("Om",19,"Male")

print(person._age)
```

```python
## Encapsulation with getter and setter method

class Person:
    def __init__(self,name,age):
        self.__name=name            
        self.__age=age   

    ## getter method for name
    def get_name(self):
        return print(self.__name)

    ## setter method for name
    def set_name(self,name):
        self.__name=name

    ## getter method for age
    def get_age(self):
        return print(self.__age)

    ## setter method for name
    def set_age(self,age):
        if age > 0:
            self.__age=age

        else:
            print("Age cannot be negative.")


person=Person("om",19)

person.get_name()
person.set_name("Sai")
person.get_name()
person.get_age()
person.set_age(-2)
person.set_age(13)
person.get_age()



           
```
