## Polymorphism

### Method Overriding

Method Overriding allow a child class to provide a specific implementation of a method that already defined in its parent class

```python
## Base class

class Animal:
    def speak(self):
        return "Sound of the animal"

## Derived Class
class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak (self):
        return "Meow"

class Cow(Animal):
    pass

def animal_speak(animal):
    print(animal.speak())
    

dog=Dog()
cat=Cat()
cow=Cow()

print(dog.speak())
print(cat.speak())
print(cow.speak())
animal_speak(dog)
```

```python
### Polymorphism with function and Method

##Base Class
from math import pi

class Shape:
    def area(self):
        return "The are of the figure"
    
##derived Class 1

class Rectangle(Shape):
    def __init__(self,lenght,width):
        self.lenght=lenght
        self.width=width

    def area(self):
        return self.width * self.lenght
    
##derived Class 1

class Triangle(Shape):
    def __init__(self,base,height):
        self.base=base
        self.height=height

    def area(self):
        return (1/2)*self.base*self.height
    
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return pi*(self.radius**2)
    
## Function that demonstrate polymorphism

def print_area(shape):
    print(f"the area is {shape.area()}")

rec=Rectangle(4,5)
tri=Triangle(4,3)
cir=Circle(6)

print_area(rec)
print_area(tri)
print_area(cir)
```

```python
## Polymorphism with abstract base classes

from abc import ABC,abstractmethod

## Define an abstract class 
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

## Derived class 1

class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

## Derived class 2

class Bike(Vehicle):
    def start_engine(self):
        return "Bike engine started"
    
def vehicle_start(vehicle):
    print(vehicle.start_engine())
    
## creat object

car=Car()
bike=Bike()

vehicle_start(car)
vehicle_start(bike)


```
