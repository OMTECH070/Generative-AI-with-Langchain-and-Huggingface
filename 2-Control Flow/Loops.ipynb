## Loops

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
## Srtring

str="Om Mahadik"
for i in str :
    print(i,end="")

```

```python
##While Loop

count=0

while count <5 :
    print("Hello")
    count+=1
```

```python


count=0

while count %2==0 :
    print("Hello")
    count+=1
```

```python
##Loop control statement

#Break
## Exits the loop permaturely

for i in range(10):
    if i==5:
        break
    print(i)
    
```

```python
#Continue
##Skips the current iteration and continues with the next

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
#Pass
##Null operation;  it does nothing

for i in range(5):
    if i ==3:
        pass
    print(i)
```

```python
## Nested Loop

for i in range(3):
    for j in range (2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples -Calculate the sum of 1st N natural numbers using while and for loop

#While Loop

N=int(input("Enter the number:"))
Total=0
x=0
while x<=N:
    Total+=(N-x)
    x+=1
print(Total)
    
```

```python
#For Loop

N=int(input("Enter the number:"))
total=0

for i in range (N+1):
    total+=i

print(i)

```

```python
## Examples - Prime numbers between 1 and 100

for i in range (1,101):
    if i >1:
        for j in range(2,i):
            if i%j==0:
                break
        else :
            print(i)
    
```
