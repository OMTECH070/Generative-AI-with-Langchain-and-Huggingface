## Valid Sudoku

```python
s=np.array([
           [5, 3, 4, 6, 7, 8, 9, 1, 2],
           [6, 7, 2, 1, 9, 5, 3, 4, 8],
           [1, 9, 8, 3, 4, 2, 5, 6, 7],
           [8, 5, 9, 7, 6, 1, 4, 2, 3],
           [4, 2, 6, 8, 5, 3, 7, 9, 1],
           [7, 1, 3, 9, 2, 4, 8, 5, 6],
           [9, 6, 1, 5, 3, 7, 2, 8, 4],
           [2, 8, 7, 4, 1, 9, 6, 3, 5],
           [3, 4, 5, 2, 8, 6, 1, 7, 9] ] )
s
```

```python
a=np.sum(s,axis=1)
for i in a:
    if i!=45:
        print("Sudoku is not valid ")
        break
else:
    print("Sudoku is  valid for rows")
```

```python
b=np.sum(s,axis=0)
for i in a:
    if i!=45:
        print("Sudoku is not valid ")
        break
else:
    print("Sudoku is  valid for colums")
```

```python

# Assume s is your 9x9 numpy array
is_valid = True  # Start by assuming it is valid

for i in range(0, 9, 3):
    for j in range(0, 9, 3):
        n = s[i:i+3, j:j+3]
        
        # If ANY box does NOT sum to 45, it is invalid
        if n.sum() != 45:
            is_valid = False
            break # Stop checking inner loop
    if not is_valid:
        break # Stop checking outer loop

# Print the final result ONCE after checking all boxes
if is_valid:
    print("Sudoku is valid")
else:
    print("Sudoku is not valid")   
```
