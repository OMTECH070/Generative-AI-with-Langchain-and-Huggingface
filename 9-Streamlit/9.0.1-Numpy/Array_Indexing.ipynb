```python
import numpy as np
```

## Numpy Indexing And Slicing Of Vector

```python
arr=np.arange(11,21)
arr
```

```python
arr[6]
```

```python
print(arr[1:5])
print(arr[:5])
print(arr[1:])
print(arr[1:5:2])
print(arr[::-1])
```

## Numpy Indexing And Slicing Of Marix

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
print(arr[3])
print(arr[4])
print(arr[4,2])
print(arr[0,0])
```

```python
## Whta if you want the matrix of 2,3 and 7,8

slice=arr[0:2,1:3]
slice
```

```python
## Slice 19,29 and 24,25 and 29,30

slice=arr[3:,3:]
slice
```

```python
arr[:,2]
```
