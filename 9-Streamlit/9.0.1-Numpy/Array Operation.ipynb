## Array Operations

Arithmetic Operation

```python
a1=np.array([1,2,3,4,5])      ## Size of array must be same
a2=np.array([6,7,8,9,10])
```

```python
a = a1 + a2
a
```

```python
a1 - a2
```

```python
a1 / a2
```

```python
a1 * a2
```

```python
a1 ** a2
```

```python
a1 // a2
a
```

Broadcasting

```python
l=[10,20,30,40]  ##In list you have if you wanted to add 10 in all the elements then you have to loop it over then add 10 and the append arrays do this in one go
arr=np.array(l)
```

```python
arr + 10
```

```python
a=np.arange(1,26).reshape(5,5)
a
```

Deep and Shallow copy

```python
a=np.arange(1,21)
a
```

```python
slice=a[:5]
slice
```

```python
slice+=10
slice
```

## Advance Array Manipulation

Stacking Array

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```
