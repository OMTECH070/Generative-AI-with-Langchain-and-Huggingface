```python
import numpy as np
import pandas as pd
```

## Finding Missing Data

```python
data = {
    'A': [1, 2, None, 4],
    'B': [None, 6, 7, 8],
    'C': [9, None, 11, 12],
    'D': [13, 14, 15, None],
    'E': [17, 18, 19, 20]
}   

df1=pd.DataFrame(data)
```

```python
df1
```

```python
df1.isna()
```

```python
df1.isna().sum()
```

```python
df1.isna().any()
```

## Removing Missing Data

```python
df1.dropna()
```

```python
df1
```

```python
df1.dropna(thresh=3)
```

## Filling The Missing Data

```python
df1
```

```python
df1.fillna(0)
```

```python
values={"A":0,"B":1,"C":2,"D":3,"E":4}
df1.fillna(value=values)
```

```python
df1.fillna(df1.mean())
```
