```python
import numpy as np
import pandas as pd
```

## Merging 2 Dataframes

```python
employees = {
    "EmployeeID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 28, 35, 27],
    "City": ["New York", "London", "Mumbai", "Sydney", "Toronto"],
}

salaries={
        "EmployeeID": [101, 102, 103, 106, 107],
        "Salary": [55000, 62000, 48000, 75000, 59000],
        "Bonus": [5000, 6000, 4000, 7000, 5500]
}
Em=pd.DataFrame(employees)
Sa=pd.DataFrame(salaries)
```

```python
Em
```

```python
Sa
```

```python
pd.merge(Em,Sa,on="EmployeeID",how='inner')
```

```python
pd.merge(Em,Sa,on="EmployeeID",how='outer')
```

```python
pd.merge(Em,Sa,on="EmployeeID",how='left')
```

```python
pd.merge(Em,Sa,on="EmployeeID",how='right')
```

## Concatination of 2 dataframes

```python
df1=pd.DataFrame({ 
    "A":["A0","A1","A2"],
    "B":["B0","B1","B2"],
    "C":["C0","C1","C2"]})


df2=pd.DataFrame({
    "A":["A3","A4","A5"],
    "B":["B3","B4","B5"],
    "C":["C3","C4","C5"]
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2] ,axis=1)
```

## Joining of 2 Dataframes

```python
df1=pd.DataFrame({
    "name":["Alice","Bob","Charlie"]
},index=[0,1,2])

df2=pd.DataFrame({
    "age":[25,30,28]
},index=[0,3,4])                    
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df1.join(df2,how='outer')
```

```python
df2.join(df1)
```
