## GroupBy

```python
data={
    "Category":["A","B","A","B","A"],
    "Store":["X","X","Y","Y","Z"],
    "Sales":[100,200,150,300,250],
    "Quantity":[10,20,15,30,25],
    "Date":pd.date_range('2026-01-01',periods=5)
}
df=pd.DataFrame(data)
```

```python
df
```

```python
## Group by category and calculate the sum of the Sales 
cat=df.groupby("Category")["Sales"].sum()
print(cat)
```

```python
## Group by Store and calculate the sum of the Sales 
cat=df.groupby("Store")["Sales"].sum()
cat
```

```python
# Group by multiple columns 
# Group by Category and Store and calculate the sum of Sales and Quantity

cat=df.groupby(["Category", "Store"])["Sales"].sum()
cat
```
