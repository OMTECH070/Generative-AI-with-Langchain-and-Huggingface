**Now, Here comes some Basic Questions**

1. Which team won the most matches?

```python
winner=df['match_winner'].value_counts()
sns.barplot(y=winner.index,x=winner.values,palette='viridis')
plt.title("Most Match Win By Team")

```

2. Toss Decision Trends

```python
df.head()
```

```python
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title("Toss Decision Trends")
```

3. Toss Winner vs Match Winner

```python
x=df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage=(x/(df["match_id"].count()))*100
percentage.round(2)
```

4. How do teams win? (Runs vs Wickets)

```python
sns.countplot(x=df['won_by'])
plt.title("won_by")
```

###  Key Player Performances

1 Most "Player of the Match" Awards

```python
winner=df['player_of_the_match'].value_counts().head()
sns.barplot(y=winner.index,x=winner.values,palette='viridis')
plt.title("player_of_the_match")
```

2 Top Scorers

```python
players=df['top_scorer'].value_counts().head(2)
sns.barplot(x=players.values,y=players.index,palette='mako')
```

```python
high=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending = False).head(2)
high
sns.barplot(high)
```

10 Best Bowling Figures

```python
df.head()
```

```python
df['highest']=df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
df['highest']=df['highest'].astype(int)
df['highest']

hih=df.groupby('best_bowling')['highest'].sum().sort_values( ascending = False).head(10)
hih
sns.barplot(x=hih.values,y=hih.index)

```
