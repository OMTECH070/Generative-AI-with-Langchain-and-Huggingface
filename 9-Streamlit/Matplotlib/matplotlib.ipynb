```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Matplotlib Basics

```python
x=np.linspace(0,5,11)
x
y=x**3
y
```

```python
plt.plot(x,y)
plt.title("Test Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

## Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.title("Test 1")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.subplot(2,2,2)
plt.plot(x,x)
plt.title("Test 2")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.subplot(2,2,3)
plt.plot(y,x)
plt.title("Test 3")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.subplot(2,2,4)
plt.plot(y,y)
plt.title("Test 4")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

## Object Oriented way

```python
fig=plt.figure(figsize=(10,8),dpi=100)
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis1=fig.add_axes([0.4,0.4,0.2,0.2])
axis1.plot(y,x)
```

## Types of Plots

```python
plt.scatter(x,y)
```

```python
plt.hist(y)
```

```python
plt.boxplot(x)
```

## Saving Plots

```python
fig=plt.figure(figsize=(10,8),dpi=100)
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
plt.savefig("Basic.png")
```
