```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## Heat map

```python
flights=sns.load_dataset('flights')
flights
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
tipscore=tips[["total_bill"	,"tip","size"]]
tipscore
```

```python
tipscore.corr()
```

```python
sns.heatmap(tipscore.corr(),annot=True)
```

## Cluster Map

```python
sns.clustermap(tipscore.corr(),annot=True)
```

## Pivot table heat map

```python
pivotflight=pd.pivot_table(flights,values="passengers",index="month",columns="year")
pivotflight
```

```python
sns.heatmap(pivotflight)
```
