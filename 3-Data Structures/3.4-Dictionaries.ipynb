## Dictionaries

Dictionaries are unordered collection of item.They store data in key-value pair,Keys must be unique and immutable,while values can be of any data type

```python
dic={}
print(type(dic))
```

```python
student={"name":"Om","age":18,"grade":100}
print(student)
```

```python
##Single key is always used
student={"name":"Om","age":18,"name":100}
print(student)
```

```python
student={"name":"Om","age":18,"grade":"A"}
print(student["grade"])
print(student["age"])

print(student.get("grade"))
print(student.get("last_name","Not Available"))

```

```python
##Modifying
student["age"]=20
student["address"]="India"

print(student)
```

```python
del student["grade"]
print(student)
```

```python
keys=student.keys()
print(keys)

values=student.values()
print(values)

items=student.items()
print(items)
```

```python
##Shallow Copy

student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="Sai"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy()
print(student)
print(student_copy1)
```

```python
student["name"]="Raj"
print(student)
print(student_copy1)
```

```python
#Keys
for keys in student.keys():
                         print(keys)
```

```python
#Values
for Value in student.values():
                         print(Value)
```

```python
#Items
for key,value in student.items():
                         print(f"{key}:{value}")
```

```python
##Nested Dictionaries
students={
    "student1":{"name":"Om","Age":18},

    "student2":{"name":"Sai","Age":12},

    "student3":{"name":"Raj","Age":22}

}
```

```python
print(students["student1"]["name"])
print(students["student3"]["Age"])
```

```python
for student_id,student_info in students.items():
    print(student_id)
    for x,y in student_info.items():
        print(f"{x}:{y}")
    print()
```

```python
##Dictionaries Comprehension
square={x:x**2 for x in range(10)}
print(square)
```

```python
##Conditional Dictionaries Comprehension
square={x:x**2 for x in range(10) if x%2==0}
print(square)
```

```python
## Practical Example 

##Use a dictionaries to count the frequency of elements in list 

numbers=[1,2,3,4,5,1,2,3,5,9,1,1,6,6,8,3,6,9,1]
frequency={}
for x in numbers:
    if x in frequency:
        frequency[x]+=1
    else:
        frequency[x]=1

print(frequency)

```

```python
##Merge 2 dictionaries into one

dic1={"a":1,"b":2}
dic2={"c":3,"d":4}

dic={**dic1,**dic2}
print(dic)

```
