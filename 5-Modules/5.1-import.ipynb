## Importing Modules in Python Modules and Packages

```python
import math 
math.sqrt(16)
```

```python
from math import sqrt,pi

print(sqrt(25))
print(pi)
```

```python
import numpy as np

np.array([1,2,3,4])
```

```python
from math import *

print(sqrt(16))
print(pi)
```

```python
from package.maths import add

add(2,6)
```

```python
from package import maths

maths.sub(9,5)
maths.add(8,9)
```
